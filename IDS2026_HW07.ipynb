**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO
plt.figure(figsize=(12, 7))

sns.regplot(data=df,
            y="total_sleep_hours",
            x="bedtime_phone_minutes",
            line_kws={"color" : "red"},
            scatter_kws={"alpha": 0.5, "color": "blue"})
plt.title("Bedtime phone usage vs Total sleep")
plt.ylabel("Total sleep (hours)")
plt.xlabel("Bedtime phone usage (minutes)")

plt.tight_layout()
plt.show()
```

**Answer:** A clear trend, where higher phone usage before bed lowers the amount of hours slept
